# 01 — Market Data and Returns

**Objective:** inspect the price data for an initial pair and measure how closely their daily returns move together.

I start with Coca-Cola (`KO`) and PepsiCo (`PEP`) because they operate in the same broad industry and provide an intuitive first pair to investigate.

```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf

TICKERS = ["KO", "PEP"]
START_DATE = "2018-01-01"
END_DATE = "2026-01-01"
```

```python
data = yf.download(
    TICKERS,
    start=START_DATE,
    end=END_DATE,
    auto_adjust=True,
    progress=False
)

prices = data["Close"].dropna()

print(prices.head())
print(f"Observations: {len(prices)}")
```

## Price comparison

I normalise both series to 1 at the start so their relative growth is easier to compare.

```python
normalised = prices / prices.iloc[0]

normalised.plot(figsize=(10, 5))
plt.title("Normalised KO and PEP Prices")
plt.ylabel("Growth of 1 unit")
plt.xlabel("Date")
plt.grid(alpha=0.3)
plt.show()
```

## Daily returns and correlation

Correlation measures how closely short-term price movements line up. It is useful as an initial check, but it does not establish a stable long-run relationship.

```python
returns = prices.pct_change().dropna()

correlation = returns["KO"].corr(returns["PEP"])

print(f"Return correlation: {correlation:.3f}")
returns.describe()
```
